```python
import re
from gensim.models import Word2Vec

# 1. Load textual document
with open("document3.txt", "r", encoding="utf-8") as file:
    text = file.read()

# 2. Clean and tokenize the document
text = re.sub(r'[^a-zA-Z\s]', '', text.lower())

sentences = []

for sentence in text.split('\n'):
    words = sentence.split()

    if words:
        sentences.append(words)

# 3. Train CBOW Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=100
)

# 4. Evaluate the model
print("Vocabulary Size:", len(model.wv))

# Display word embedding and similar words
word = "virus"

if word in model.wv:
    print("Word Vector:", model.wv[word])

    print("Similar Words:")
    print(model.wv.most_similar(word, topn=5))
else:
    print(word, "not found in vocabulary")
```
