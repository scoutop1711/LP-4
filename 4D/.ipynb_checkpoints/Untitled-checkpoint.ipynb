```python
import pandas as pd
import re
import numpy as np

from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
```

```python
df = pd.read_csv("twitter_dataset.csv")

df.head()
```

```python
stop_words = set(ENGLISH_STOP_WORDS) - {"not", "no", "never"}
```

```python
def preprocess(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+|@\w+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)

    words = text.split()

    clean_words = []
    for word in words:
        if word not in stop_words:
            clean_words.append(word)

    return clean_words
```

```python
df = df.dropna(subset=["Text"]).copy()

tokens_list = []
for text in df["Text"]:
    tokens_list.append(preprocess(text))

df["tokens"] = tokens_list
```

```python
df = df[df["tokens"].map(len) > 0].reset_index(drop=True)
```

```python
model = Word2Vec(
    sentences=df["tokens"].tolist(),
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=20
)
```

```python
def document_vector(words):
    vectors = []
    for word in words:
        if word in model.wv:
            vectors.append(model.wv[word])

    if len(vectors) == 0:
        return np.zeros(100)

    return np.mean(vectors, axis=0)
```

```python
tweet_vectors = []
for words in df["tokens"]:
    tweet_vectors.append(document_vector(words))

tweet_vectors = np.array(tweet_vectors)
```

```python
user_input = input("Enter your search text: ")
user_tokens = preprocess(user_input)
user_vector = document_vector(user_tokens)
```

```python
known_words = False
for word in user_tokens:
    if word in model.wv:
        known_words = True
        break
```

```python
if known_words == False:
    print("No known words found. Please enter different text.")
else:
    similarities = cosine_similarity(
        user_vector.reshape(1, -1),
        tweet_vectors
    )[0]
```

```python
top_indices = np.argsort(similarities)[::-1][:5]
```

```python
print("\nRecommended Tweets:\n")

for i, idx in enumerate(top_indices):
    print("Recommendation", i + 1)
    print("Tweet:", df.iloc[idx]["Text"])
    print("Similarity:", round(similarities[idx], 4))
    print("-" * 50)
```
