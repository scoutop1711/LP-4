# PS 3B
Implement Industrial Equipment Anomaly Detection on the MVTec AD Kaggle
dataset using Autoencoder and build the model by using the following steps:
Steps:
  
  • Preprocess images (resize, normalize) for input into a convolutional autoencoder.
  
  • Build a convolutional autoencoder (CNN-based encoder-decoder) to learn the
representations of normal object images.
  
  • Use reconstruction errors in pixel values to detect anomalies in test images.
  
  Dataset: MVTec AD is an anomaly detection dataset for industrial defect detection,
containing images of normal and defective objects.
  
  Goal: Detect defects in objects by training an image-based autoencoder on images of
defect-free objects.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

from tensorflow.keras import models, layers, optimizers
```

```python
# LOADING DATASET
df = pd.read_csv("ecg.csv", header = None)
df.head()
```

```python
print(df.shape)
print(df.iloc[:, -1].isnull().sum().sum())
```

```python
# SEPARATE X AND Y
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
# NORMALIZE DATA
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
# SPLIT THE DATASET
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size = 0.2, random_state = 42
)
```

```python
# SEPARATING NORMAL AND ABNORMAL ECG
X_train = X_train[y_train == 0]
```

```python
# DEFINING AUTOENCODER ARCHITECTURE

# ENCODER
encoder = models.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(64, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(16, activation="relu"),
    layers.Dense(8, activation="relu")
])

# DECODER
decoder = models.Sequential([
    layers.Input(shape=(8,)),
    layers.Dense(16, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(X_train.shape[1], activation="sigmoid")
])

# AUTOENCODER
model = models.Sequential([encoder, decoder])

model.summary()
encoder.summary()
decoder.summary()
```

```python
# COMPILE THE MODEL
model.compile(
    optimizer = 'adam',
    loss = "mse",
    metrics = ["mae"]
)
```

```python
# TRAIN THE MODEL
history = model.fit(
    X_train, X_train,
    epochs = 10,
    batch_size = 256,
    validation_split = 0.1
)
```

```python
# CALCULATE RECONSTRUCTION ERROR
train_predictions = model.predict(X_train)

train_errors = np.mean(
    np.square(X_train - train_predictions), axis = 1 
)
```

```python
# SETTING THE THRESHOLD TO 95TH PERCENTILE
threshold = np.percentile(train_errors, 90 )

print("Threshold: ", threshold)
```

```python
# DETECT FRAUDENT TRANSACTIONS
test_predictions = model.predict(X_test)

test_errors = np.mean(
    np.square(X_test - test_predictions), axis = 1 
)

y_pred = (test_errors > threshold).astype(int)

print("Reconstruction Errors:")
print(test_errors[:20])

print("Predicted Classes:")
print(y_pred[:20])
```

```python
# EVALUATE MODEL PERFORMANCE
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
