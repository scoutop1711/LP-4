Complete program

```python
import pandas as pd
import re
from gensim.models import Word2Vec

# 1. Load Twitter dataset
data = pd.read_csv("twitter_dataset.csv")

# 2. Clean and tokenize tweets
sentences = []

for tweet in data["Text"].dropna():
    tweet = re.sub(r'[^a-zA-Z\s]', '', str(tweet).lower())
    words = tweet.split()

    if words:
        sentences.append(words)

# 3. Train CBOW Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=10
)

# 4. Evaluate the model
print("Vocabulary Size:", len(model.wv))

# Display word embedding and similar words
word = "government"

if word in model.wv:
    print("Word Vector:", model.wv[word])

    print("Similar Words:")
    print(model.wv.most_similar(word, topn=5))
else:
    print(word, "not found in vocabulary")
```
